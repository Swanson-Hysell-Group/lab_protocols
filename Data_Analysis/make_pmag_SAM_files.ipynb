# Notebook for quickly generate CIT format magnetometer specimen files
- use this script if you want to skip dec corrections, sun corrections, etc and just create a file with a given set of orientation data

```python
import os
import pandas as pd
```

```python
all_core_orientations = pd.read_csv('./U1555_coords.csv')
all_core_orientations
```

```python
for i in range(all_core_orientations.shape[0]):
    this_sample_name = all_core_orientations.iloc[i]['sample_name']
    this_comment = all_core_orientations.iloc[i]['comment']
    this_strat_level = all_core_orientations.iloc[i]['strat_level']
    this_core_strike = all_core_orientations.iloc[i]['magnetic_core_strike']
    this_core_dip = all_core_orientations.iloc[i]['core_dip']
    this_bedding_strike = all_core_orientations.iloc[i]['bedding_strike']
    this_bedding_dip = all_core_orientations.iloc[i]['bedding_dip']
    this_mass = all_core_orientations.iloc[i]['mass']


    # write file
    if this_core_strike is None or pd.isna(this_core_strike):
        continue
    with open(f'./U1555_specimen_files/{this_sample_name}', 'w') as f:
        f.write(f'{this_sample_name} mag compass orientation\r\n')
        f.write('{:7.1f}'.format(this_strat_level) + '{:6.1f}'.format(this_core_strike) +  '{:6.1f}'.format(this_core_dip) + '{:6.1f}'.format(this_bedding_strike) + '{:6.1f}'.format(this_bedding_dip) + '{:6.1f}'.format(this_mass) + '\r\n')
    f.close()
```
